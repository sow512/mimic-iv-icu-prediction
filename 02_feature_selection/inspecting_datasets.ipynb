```python
# looking at 300 features

import pandas as pd
from pathlib import Path

# ============================================================
# 1. FILE PATHS
# ============================================================

DATASET_200_PATH = Path(
    '. Group Project/Post mid-term/200_Feature_Dataset.xlsx'
)

DATASET_75_PATH = DATASET_3_PATH  # Already defined in your notebook

# ============================================================
# 2. LOAD 200-FEATURE DATASET
# ============================================================

# Feature dictionary (fast to load)
features_200_df = pd.read_excel(
    DATASET_200_PATH,
    sheet_name="Feature Dictionary"
)

# Extract the 200 feature names
features_200 = set(
    features_200_df["Selected feature"]
    .dropna()
    .astype(str)
)

print("200-feature dataset loaded.")
print("Number of selected features:", len(features_200))

# ============================================================
# 3. LOAD 75-FEATURE DATASET
# ============================================================

df_75 = pd.read_excel(
    DATASET_75_PATH,
    sheet_name="Final_75_Features"
)

exclude_75 = {
    "subject_id",
    "hadm_id",
    "dataset_split",
    "icu_transfer"
}

features_75 = set(df_75.columns) - exclude_75

print("\n75-feature dataset loaded.")
print("Dataset shape:", df_75.shape)
print("Number of selected features:", len(features_75))

# ============================================================
# 4. COMPARE FEATURE SETS
# ============================================================

common_features = features_200 & features_75

only_75 = features_75 - features_200
only_200 = features_200 - features_75

print("\n" + "=" * 60)
print("FEATURE COMPARISON")
print("=" * 60)

print("Features in 200-feature dataset:", len(features_200))
print("Features in 75-feature dataset:", len(features_75))
print("Common features:", len(common_features))

print("\nIs the 75-feature set a subset of the 200-feature set?")
print(features_75.issubset(features_200))

print("\nFeatures present in 75 but absent from 200:")
print(sorted(only_75))

print("\nFeatures present in 200 but absent from 75:")
print(sorted(only_200))
```

```python
# Load actual patient-level data
train_df = pd.read_excel(
    DATASET_200_PATH,
    sheet_name="Training Data"
)

test_df = pd.read_excel(
    DATASET_200_PATH,
    sheet_name="Testing Data"
)

print("200-feature training shape:", train_df.shape)
print("200-feature testing shape:", test_df.shape)

print("\nTraining patients:", train_df["subject_id"].nunique())
print("Testing patients:", test_df["subject_id"].nunique())
```

```python
# ============================================================
# COMPARE 75-FEATURE AND 200-FEATURE DATASET SPLITS
# ============================================================

train_75 = df_75[
    df_75["dataset_split"] == "train"
].copy()

test_75 = df_75[
    df_75["dataset_split"] == "test"
].copy()

print("75-feature training rows:", len(train_75))
print("75-feature testing rows:", len(test_75))

# Compare admission IDs
train_200_ids = set(train_df["hadm_id"])
test_200_ids = set(test_df["hadm_id"])

train_75_ids = set(train_75["hadm_id"])
test_75_ids = set(test_75["hadm_id"])

print("\nTRAINING SPLIT")
print("Shared admissions:", len(train_200_ids & train_75_ids))
print("Only in 200-feature training:", len(train_200_ids - train_75_ids))
print("Only in 75-feature training:", len(train_75_ids - train_200_ids))

print("\nTESTING SPLIT")
print("Shared admissions:", len(test_200_ids & test_75_ids))
print("Only in 200-feature testing:", len(test_200_ids - test_75_ids))
print("Only in 75-feature testing:", len(test_75_ids - test_200_ids))

# Check patient overlap within 75-feature dataset
shared_patients = (
    set(train_75["subject_id"]) &
    set(test_75["subject_id"])
)

print("\nShared patients across 75-feature train/test:",
      len(shared_patients))
```

```python
import pandas as pd
import numpy as np

# Combine original 200-feature training/testing data
df_200_all = pd.concat(
    [train_df, test_df],
    ignore_index=True
)

# Compare ICU-transfer labels
labels = df_200_all[
    ["hadm_id", "icu_transfer"]
].merge(
    df_75[["hadm_id", "icu_transfer"]],
    on="hadm_id",
    suffixes=("_200", "_75"),
    validate="one_to_one"
)

print("Conflicting target labels:",
      (labels["icu_transfer_200"] !=
       labels["icu_transfer_75"]).sum())

# Compare shared feature values
shared_features = sorted(features_200 & features_75)

merged = df_200_all[
    ["hadm_id"] + shared_features
].merge(
    df_75[["hadm_id"] + shared_features],
    on="hadm_id",
    suffixes=("_200", "_75"),
    validate="one_to_one"
)

mismatches = {}

for feature in shared_features:
    a = pd.to_numeric(
        merged[f"{feature}_200"], errors="coerce"
    )
    b = pd.to_numeric(
        merged[f"{feature}_75"], errors="coerce"
    )

    mismatches[feature] = (~np.isclose(
        a, b,
        rtol=1e-5,
        atol=1e-6,
        equal_nan=True
    )).sum()

mismatches = pd.Series(mismatches)

print("Shared features with mismatched values:",
      (mismatches > 0).sum())

display(
    mismatches[mismatches > 0]
    .sort_values(ascending=False)
)
```

```python
# ============================================================
# CHARACTERIZE FEATURE MISMATCHES
# ============================================================

import numpy as np
import pandas as pd

mismatch_details = []

for feature in shared_features:

    a = pd.to_numeric(
        merged[f"{feature}_200"],
        errors="coerce"
    )

    b = pd.to_numeric(
        merged[f"{feature}_75"],
        errors="coerce"
    )

    both_present = a.notna() & b.notna()

    one_missing = a.isna() ^ b.isna()

    different_values = (
        both_present &
        ~pd.Series(
            np.isclose(
                a.fillna(0),
                b.fillna(0),
                rtol=1e-5,
                atol=1e-6
            ),
            index=a.index
        )
    )

    mismatch_details.append({
        "Feature": feature,
        "Different_Numeric_Values": int(different_values.sum()),
        "Missingness_Disagreements": int(one_missing.sum()),
        "Total_Mismatches": int(
            different_values.sum() + one_missing.sum()
        ),
        "Mismatch_Percent": round(
            100 * (
                different_values.sum() + one_missing.sum()
            ) / len(merged),
            2
        )
    })

mismatch_report = pd.DataFrame(mismatch_details)

mismatch_report = mismatch_report.sort_values(
    "Total_Mismatches",
    ascending=False
)

display(mismatch_report.head(25))
```
